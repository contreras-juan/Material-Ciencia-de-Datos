Grafica `loss` vs `val_loss` y `accuracy` vs `val_accuracy`.


```python
def plot_training_curves(history, metrics=("loss", "accuracy")):
    """Grafica curvas de entrenamiento y validación."""
    history_dict = history.history if hasattr(history, "history") else history
    n = len(metrics)
    plt.figure(figsize=(6 * n, 4))
    for i, metric in enumerate(metrics):
        plt.subplot(1, n, i + 1)
        plt.plot(history_dict[metric], label=f"train {metric}")
        plt.plot(history_dict[f"val_{metric}"], label=f"val {metric}")
        plt.xlabel("Epochs")
        plt.ylabel(metric)
        plt.title(metric)
        plt.legend()
        plt.grid(True)
    plt.tight_layout()
    plt.show()

# Escribe tu código aquí
# Entrena un modelo propenso a sobreajuste y llama plot_training_curves(...)
```
